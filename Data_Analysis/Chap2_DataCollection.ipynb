```python
# 딕셔너리를 json 형태로 구성
d = {"name": "혼자 공부하는 데이터 분석"}
print(d['name'])
```

```python
import json
```

```python
# 딕셔너리 d에 한글이 포함되어 있어, 원래 저장된 문자를 그대로 출력하도록 설정함
d_str = json.dumps(d, ensure_ascii=False)
print(d_str)
```

```python
# d_str이 문자열 객체인지 확인
print(type(d_str))
```

### 웹 기반 API에서 데이터 전송 시 파이썬 객체를 문자열로 바꾸어 전송하는 이유:
> #### 웹 기반 API가 사용하는 HTTP 프로토콜이 텍스트 기반이기 때문에
- 직렬화(serialization): 프로그램 상의 객체를 저장하거나 읽을 수 있는 형태로 변환
- 역직렬화(deserialization): 직렬화된 정보를 다시 프로그램에서 실행 가능한 객체로 변환

### XML 데이터 다루기

```python
x_str = """
<book>
    <name>혼자 공부하는 데이터 분석</name>
    <author>박해선</author>
    <year>2022</year>
</book>
"""
```

```python
import xml.etree.ElementTree as et
book = et.fromstring(x_str)
```

```python
print(type(book))
```

```python
print(book.tag)
```

```python
book_childs = list(book)
print(book_childs)
```

```python
name, author, year = book_childs
print(name.text)
print(author.text)
print(year.text)
```

#### findtext() 메서드를 사용해야 하는 이유

- XML은 자식 엘리먼트 순서를 항상 일정하게 보장하지 않음
    - findtext() 메서드 사용 시 자식 엘리먼트를 탐색하여 자동으로 텍스트를 반환함

```python
name = book.findtext('name')
author = book.findtext('author')
year = book.findtext('year')
print(name)
print(author)
print(year)
```

```python
x2_str = """
<books>
    <book>
        <name>혼자 공부하는 데이터 분석</name>
        <author>박해선</author>
        <year>2022</year>
    </book>
    <book>
        <name>혼자 공부하는 머신러닝+딥러닝</name>
        <author>박해선</author>
        <year>2020</year>
    </book>
</books>
"""
```

```python
books = et.fromstring(x2_str)
print(books.tag)
```

```python
for book in books.findall('book'):
    name = book.findtext('name')
    author = book.findtext('author')
    year = book.findtext('year')

    print(name)
    print(author)
    print(year)
    print()
```
